# 01 - Exploration des locations au Sénégal
Ce notebook documente un jeu synthétique reproductible. La logique réutilisable reste dans `src/`.

```python
from pathlib import Path
import pandas as pd
from senegal_rental_price.data.generate import generate_dataset
from senegal_rental_price.data.preprocessing import clean_data
```

## 1. Provenance
Option 2 de l'énoncé : données synthétiques, graine 42.

```python
raw_path = Path('../data/raw/locations_senegal.csv')
raw = pd.read_csv(raw_path) if raw_path.exists() else generate_dataset(seed=42)
raw.shape, raw.head()
```

## 2. Qualité et nettoyage

```python
quality = pd.DataFrame({'dtype': raw.dtypes.astype(str), 'missing': raw.isna().sum(), 'unique': raw.nunique()})
quality
```

```python
data = clean_data(raw)
data.describe(include='all').T
```

## 3. Distribution des prix

```python
data.groupby('ville')['prix_loyer_mensuel'].agg(['count', 'median', 'mean']).round(0)
```

```python
data.groupby('type_bien')['prix_loyer_mensuel'].median().sort_values().plot(kind='barh', title='Loyer médian synthétique par type de bien');
```

## 4. Relation surface-prix

```python
data.plot.scatter(x='surface_m2', y='prix_loyer_mensuel', alpha=.3, title='Surface et loyer synthétique');
```

## 5. Conclusion
- Dakar et les quartiers premium portent une prime volontairement simulée.
- La surface et les équipements augmentent le prix.
- Ces tendances valident la cohérence du générateur, pas le marché réel.
- L'entraînement est exécuté depuis le package avec Hydra et MLflow.
